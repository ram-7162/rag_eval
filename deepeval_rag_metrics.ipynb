# Retriever

### (1) ContextualPrecision

- Out of all retrived documents , how many are relevant for answer genration


- We calculate it we compare each chunk with actual answer and find out whether given chunk contain info relevant to actual answer if yes then that chunk as relevant else no, let say we get 8 retrieved chunk and out of which 6 get relevant to actual answer then precison is 6/8(basic way to calculate precison, since in advance calculation order of chunks also matters)
- We consider chunk relevant even if chunk contains noise along with some info match with actual answer

- We calculate the precision with basic formula if want to study more visit https://deepeval.com/docs/metrics-contextual-precision 

### (2) ContextualRecall

- Out of all needed documents how many are retrieved successfully

- We calculate it as first we break actual context in atomic statments then we check that how many atomic statment exist in any of chunk retrived , let say our atomic statements is 3 and chunk are 5 and our 2 atomic statements match with atleat one chunk(same or different) then our recall is 2/3

- We consider chunk relevant even if chunk contains noise along with some info match with atomic statements

# Generator

### (1) Faithfulness in generator

- Faithfulness means is generated answer only answer using the retrieved context , if answer is wrong still faithfulness may be high when retrieved context is wrong

### (2) AnswerRelevancy in generator 

- AnswerRelevancy check whether generated answer is relevant to query, it did not depend over context , a wrong retrieved context may have high answer relevancy
